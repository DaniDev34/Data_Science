## 2. Lambda

```python
# Lambda basicas: cuadrado, suma, maximo


cuadrado = lambda x: x**2 
suma = lambda x, y: x + y  
maximo = lambda a, b: a if a > b else b

print("el cuadrado de x es:", cuadrado(5))
print("la suma de 5 y 7 es:", suma(5, 7))
print("el maximo de 5 y 12:", maximo(5, 12))
# Completar: cuadrado = lambda ... ; suma = lambda ... ; maximo = lambda ...
```

```python
# sorted() con lambda: ordenar por edad y por nombre (desc)
personas = [("Ana", 25), ("Carlos", 30), ("Beatriz", 22)]
# Completar: sorted(personas, key=...)
sorted(personas, key= lambda x: (x[1], x[0]))

```

```python
# max() y min() con lambda: mayor y menor edad
personas = [{"nombre": "Ana", "edad": 25}, {"nombre": "Carlos", "edad": 30}, {"nombre": "Daniel", "edad": 18}]
# Completar: max(..., key=...), min(..., key=...)
mayor = max(personas, key=lambda m: m["edad"])
menor = min(personas, key=lambda m: m["edad"])

print("es mayor:", mayor)
print("es mayor:", menor)
```

## 3. NumPy

```python
# Sin NumPy: suma con bucle for
a = [1, 2, 3, 4, 5]
b = [10, 20, 30, 40, 50]
# Completar: bucle for para sumar elementos

resultado = []

for i in range(len(a)):
    suma = a[i] + b[i]
    resultado.append(suma)

print(resultado)
```

```python
import numpy as np
# Con NumPy: operaciones +, -, *, /
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])
# Completar: a+b, a-b, a*b, b/a

resultado = a + b 
maximo = np.max(a)

print(resultado)
print(maximo)
```

```python
# Funciones NumPy: sqrt, mean, std, median
a = np.array([1, 2, 3, 4, 5])

promedio = np.mean(a)
desviacion = np.std(a)
raiz = np.sqrt(a)
mediana = np.median(a)

print("promedio: ", promedio, "desviacion: ", desviacion, "raiz: ", raiz, "mediana: ", mediana)
# Completar: np.sqrt(a), np.mean(a), np.std(a), np.median(a)
```
